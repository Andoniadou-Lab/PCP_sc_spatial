#  KI67 percentage across samples - single-cell

```python
import numpy as np
import pandas as pd
from scipy.sparse import issparse


X = adata_sc[:, "MKI67"].X
mki67 = (X.toarray() if issparse(X) else np.asarray(X)).ravel()

df = pd.DataFrame({
    "patient": adata_sc.obs["patient"].values,
    "broad_cluster": adata_sc.obs["broad_cluster"].values,
    "mki67_pos": mki67 > 0,
})

# Pct MKI67+ per cell type per patient
pct = (
    df.groupby(["patient", "broad_cluster"])["mki67_pos"]
    .mean()
    .mul(100)
    .reset_index(name="pct_mki67_pos")
)

# Mean and SD across patients for each cell type
summary = (
    pct.groupby("broad_cluster")["pct_mki67_pos"]
    .agg(["mean", "std"])
    .rename(columns={"mean": "mean_pct", "std": "sd_pct"})
)

summary
```

```python
import matplotlib.pyplot as plt

cell_types = ["Stem-like cells", "Epithelial cells"]
data = [pct.loc[pct["broad_cluster"] == ct, "pct_mki67_pos"].values for ct in cell_types]
#remove nans
data = [d[~np.isnan(d)] for d in data]
fig, ax = plt.subplots(figsize=(5, 5))
bp = ax.boxplot(data, labels=cell_types, showfliers=False, widths=0.5)

# overlay individual patient points (jittered)
import numpy as np
for i, vals in enumerate(data, start=1):
    x = np.random.normal(i, 0.04, size=len(vals))
    ax.scatter(x, vals, alpha=0.7, s=20, color="black", zorder=3)

ax.set_ylabel("% MKI67+ cells (per patient)")
ax.set_title("MKI67+ proportion per patient")
plt.tight_layout()
plt.show()
```

# KI67 in Xenium data Supp Fig 3A

```python
from scipy import stats

# Pivot to patient x cell type, paired on patient
wide = pct.pivot(index="patient", columns="broad_cluster", values="pct_mki67_pos")

# Keep patients with both stem and epithelial measured
paired = wide[["Stem-like cells", "Epithelial cells"]].dropna()

t_stat, p_val = stats.ttest_rel(paired["Stem-like cells"], paired["Epithelial cells"])

print(f"n patients (paired): {len(paired)}")
print(f"Stem mean:       {paired['Stem-like cells'].mean():.2f}%")
print(f"Epithelial mean: {paired['Epithelial cells'].mean():.2f}%")
print(f"Mean difference: {(paired['Stem-like cells'] - paired['Epithelial cells']).mean():.2f}%")
print(f"t = {t_stat:.3f}, p = {p_val:.4g}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

cell_types = ["Stem-like cells", "Epithelial cells"]

# Paired data: one row per patient with both cell types present
wide = pct.pivot(index="patient", columns="broad_cluster", values="pct_mki67_pos")
paired = wide[cell_types].dropna()

data = [paired[ct].values for ct in cell_types]

fig, ax = plt.subplots(figsize=(3, 3))
ax.boxplot(data, labels=cell_types, showfliers=False, widths=0.5)

# connect each patient across the two cell types
for _, row in paired.iterrows():
    ax.plot([1, 2], [row[cell_types[0]], row[cell_types[1]]],
            color="gray", alpha=0.5, zorder=2)

# overlay points at fixed x (no jitter)
for i, vals in enumerate(data, start=1):
    ax.scatter(np.full(len(vals), i), vals, s=20, color="black", zorder=3)

ax.set_ylabel("% MKI67+ cells (per patient)")
ax.set_title("MKI67+ proportion per patient")
plt.tight_layout()
#save svg and png
plt.savefig("/content/drive/MyDrive/human_pcp/boxplot_mki67_per_patient.svg", format="svg",bbox_inches="tight")
plt.savefig("/content/drive/MyDrive/human_pcp/boxplot_mki67_per_patient.png", dpi=800,bbox_inches="tight")
plt.show()
```
